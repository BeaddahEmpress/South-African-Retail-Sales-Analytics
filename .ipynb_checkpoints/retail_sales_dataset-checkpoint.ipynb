```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

provinces = [
    "Gauteng",
    "Western Cape",
    "Eastern Cape",
    "KwaZulu-Natal",
    "Free State",
    "Limpopo",
    "Mpumalanga",
    "North West",
    "Northern Cape"
]

categories = {
    "Groceries": ["Bread", "Milk", "Rice", "Cooking Oil"],
    "Electronics": ["Headphones", "Phone Charger", "Smart Watch"],
    "Clothing": ["T-Shirt", "Jeans", "Jacket"],
    "Home": ["Kettle", "Iron", "Microwave"],
    "Beauty": ["Shampoo", "Lotion", "Perfume"]
}

payment_methods = [
    "Cash",
    "Card",
    "EFT",
    "Mobile Payment"
]

customer_types = [
    "New",
    "Returning"
]

data = []

for i in range(1, n + 1):

    category = np.random.choice(list(categories.keys()))
    product = np.random.choice(categories[category])

    quantity = np.random.randint(1, 6)
    unit_price = round(np.random.uniform(20, 1500), 2)

    data.append([
        i,
        pd.Timestamp('2024-01-01') +
        pd.to_timedelta(np.random.randint(0, 365), unit='D'),
        np.random.choice(provinces),
        category,
        product,
        np.random.choice(customer_types),
        np.random.choice(payment_methods),
        quantity,
        unit_price,
        round(quantity * unit_price, 2)
    ])

df = pd.DataFrame(
    data,
    columns=[
        "sale_id",
        "sale_date",
        "province",
        "product_category",
        "product_name",
        "customer_type",
        "payment_method",
        "quantity",
        "unit_price",
        "total_sales"
    ]
)

df.head()
```

```python
df.to_csv(
    "south_african_retail_sales.csv",
    index=False
)

print("Dataset saved successfully!")
```
